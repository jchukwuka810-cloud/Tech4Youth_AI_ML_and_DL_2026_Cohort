```python
def perceptron(inputs, weights, bias):
    #step 1: multiply each input by its weight and add them up
    total = bias
    for w,x in zip(inputs, weights):
     total = total + x * w
     
     #step 2: activation rule
     if total > 0:
         return 1
     else:
         return 0
```

```python
# Test it with the same example: weather good, no money, friend coming
print(perceptron([1, 0, 1], [2, 3, 1], -3))

# Now with money
print(perceptron([1, 1, 1], [2, 3, 1], -3))
```
